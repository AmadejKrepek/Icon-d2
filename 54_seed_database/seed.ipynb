```python
import psycopg2
import uuid
from dotenv import load_dotenv
import os

# Load environment variables from .env
load_dotenv()

def seed_tables():
    try:
        # Access the environment variables
        DB_USERNAME = os.getenv("DB_USERNAME")
        DB_PASSWORD = os.getenv("DB_PASSWORD")
        DB_HOST = os.getenv("DB_HOST")
        DB_PORT = os.getenv("DB_PORT")
        DB_NAME = os.getenv("DB_NAME")

        # Establish a connection to the PostgreSQL database
        conn = psycopg2.connect(
            host=DB_HOST,
            port=DB_PORT,
            database=DB_NAME,
            user=DB_USERNAME,
            password=DB_PASSWORD
        )

        # Create a cursor object
        cursor = conn.cursor()

        # Seed the provider table
        cursor.execute("""
            INSERT INTO provider (id, name) VALUES
            (uuid_generate_v4(), 'DWD'),
            (uuid_generate_v4(), 'ARSO')
        """)

        # Seed the model table
        cursor.execute("""
            INSERT INTO model (id, name, last_update, status, provider_id) VALUES
            (uuid_generate_v4(), 'IconD2', NULL, NULL, (SELECT id FROM provider WHERE name = 'DWD')),
            (uuid_generate_v4(), 'Aladin', NULL, NULL, (SELECT id FROM provider WHERE name = 'ARSO'))
        """)

        # Commit the changes
        conn.commit()

        # Close the cursor and the connection
        cursor.close()
        conn.close()

        print("Tables seeded successfully.")

    except Exception as e:
        print(f"Error: {e}")

if __name__ == "__main__":
    seed_tables()
```
