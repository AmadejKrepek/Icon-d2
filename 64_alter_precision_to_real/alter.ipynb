```python
import psycopg2
from dotenv import load_dotenv
import os

# Load environment variables from .env
load_dotenv()

def convert_double_precision_to_real():
    try:
        # Access the environment variables
        DB_USERNAME = os.getenv("DB_USERNAME")
        DB_PASSWORD = os.getenv("DB_PASSWORD")
        DB_HOST = os.getenv("DB_HOST")
        DB_PORT = os.getenv("DB_PORT")
        DB_NAME = os.getenv("DB_NAME")

        # Establish a connection to the PostgreSQL database
        conn = psycopg2.connect(
            host=DB_HOST,
            port=DB_PORT,
            database=DB_NAME,
            user=DB_USERNAME,
            password=DB_PASSWORD
        )

        # Create a cursor object
        cursor = conn.cursor()

        # Define the SQL statement to alter the column type
        alter_column_sql = """
            ALTER TABLE total_precipitation
            ALTER COLUMN "Total Precipitation" TYPE REAL[][];
        """

        # Execute the SQL statement to change the column type
        cursor.execute(alter_column_sql)

        # Commit the changes
        conn.commit()

        # Close the cursor and the connection
        cursor.close()
        conn.close()

        print("Conversion completed successfully.")

    except Exception as e:
        print(f"Error: {e}")

if __name__ == "__main__":
    convert_double_precision_to_real()
```
