```python
import requests
import bz2
import re
import os

# URL of the base directory
base_url = "https://opendata.dwd.de/weather/nwp/"
file_path = "content.log.bz2"
output_folder = "variable_folders"

# Create the output folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)

# Download the content.log.bz2 file
response = requests.get(base_url + file_path)
with open(file_path, "wb") as f:
    f.write(response.content)

# Extract the content.log.bz2 file
with open(file_path, "rb") as f:
    data = bz2.decompress(f.read()).decode("utf-8")

# Use regular expressions to extract filenames and variable names
filename_pattern = r'(\w+)_(germany|europe)_regular-lat-lon_single-level_\d+_\d+_2d_(\w+)\.grib2'
filename_matches = re.findall(filename_pattern, data)

# Create a dictionary to store variable names for each model
model_variable_dict = {}

# Iterate through filename matches and store variable names for each model
for model, region, variable in filename_matches:
    if model not in model_variable_dict:
        model_variable_dict[model] = set()
    model_variable_dict[model].add(variable)

# Write the variables for each model to separate folders
for model, variables in model_variable_dict.items():
    model_folder = os.path.join(output_folder, model)
    os.makedirs(model_folder, exist_ok=True)
    
    variable_list_file = os.path.join(model_folder, f"{model}_variables.txt")
    with open(variable_list_file, "w") as f:
        f.write("List of variable names:\n")
        for variable in variables:
            f.write(variable + "\n")

    print(f"{model} variables written to {variable_list_file}. Count: {len(variables)}")

print("All variable lists written to separate folders.")
```
