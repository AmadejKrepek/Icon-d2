```python
import requests
import bz2
import re

# URL of the base directory
base_url = "https://opendata.dwd.de/weather/nwp/"
file_path = "content.log.bz2"
output_file = "variable_names.txt"

# Download the content.log.bz2 file
response = requests.get(base_url + file_path)
with open(file_path, "wb") as f:
    f.write(response.content)

# Extract the content.log.bz2 file
with open(file_path, "rb") as f:
    data = bz2.decompress(f.read()).decode("utf-8")

# Use regular expressions to extract filenames and variable names
filename_pattern = r'icon-d2_germany_regular-lat-lon_single-level_\d+_\d+_2d_(\w+)\.grib2'
filename_matches = re.findall(filename_pattern, data)
variable_names = set(filename_matches)

# Write the list of unique variable names to a text file
with open(output_file, "w") as f:
    f.write("List of variable names:\n")
    for variable_name in variable_names:
        f.write(variable_name + "\n")

print(f"Variable names written to {output_file}.")
```
