```python
import numpy as np
import imageio
import csv

# Load the radar image
image_path = '../45_hr_kompozit_download/data/2023/08/30/meteo_hr_radar_stat_kompozit_2023_08_30_10_18_26.jpg'
radar_image = imageio.imread(image_path)

# Get the dimensions of the radar image
height, width = radar_image.shape[:2]

# Initialize a dictionary to store unique values and their frequencies
unique_values = {}

# Loop through each pixel in the radar image and extract data
for y in range(height):
    for x in range(width):
        # Get the value at the current pixel
        value = tuple(radar_image[y, x])
        
        # Store value and its frequency in the dictionary
        if value in unique_values:
            unique_values[value] += 1
        else:
            unique_values[value] = 1

# Write unique values and their frequencies to a CSV file
csv_filename = 'unique_values_and_frequencies.csv'
with open(csv_filename, 'w', newline='') as csvfile:
    csv_writer = csv.writer(csvfile)
    csv_writer.writerow(['Red', 'Green', 'Blue', 'Frequency'])  # Write header
    for value, frequency in unique_values.items():
        red, green, blue = value
        csv_writer.writerow([red, green, blue, frequency])

print(f"Unique values and frequencies written to {csv_filename}")
```
