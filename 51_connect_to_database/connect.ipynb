```python
import psycopg2
import os
from dotenv import load_dotenv

# Load environment variables from .env
load_dotenv()

def check_postgres_connection(host, port, database, user, password):
    try:
        # Establish a connection to the PostgreSQL database
        conn = psycopg2.connect(
            host=host,
            port=port,
            database=database,
            user=user,
            password=password
        )

        # Check if the connection is successful
        if conn.status == psycopg2.extensions.STATUS_READY:
            print("Connection to PostgreSQL is successful.")
        else:
            print("Connection to PostgreSQL is not ready.")
        
        # Close the database connection
        conn.close()
    except Exception as e:
        print(f"Error: {e}")

# Access the environment variables
DB_USERNAME = os.getenv("DB_USERNAME")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")

check_postgres_connection(DB_HOST, DB_PORT, DB_NAME, DB_USERNAME, DB_PASSWORD)
```
