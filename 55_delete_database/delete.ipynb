```python
import psycopg2
from dotenv import load_dotenv
import os

# Load environment variables from .env
load_dotenv()

def delete_tables():
    try:
        # Access the environment variables
        DB_USERNAME = os.getenv("DB_USERNAME")
        DB_PASSWORD = os.getenv("DB_PASSWORD")
        DB_HOST = os.getenv("DB_HOST")
        DB_PORT = os.getenv("DB_PORT")
        DB_NAME = os.getenv("DB_NAME")

        # Establish a connection to the PostgreSQL database
        conn = psycopg2.connect(
            host=DB_HOST,
            port=DB_PORT,
            database=DB_NAME,
            user=DB_USERNAME,
            password=DB_PASSWORD
        )

        # Create a cursor object
        cursor = conn.cursor()

        # Delete parameter-specific tables
        cursor.execute("""
            DO $$ 
            DECLARE 
                tabname text; 
            BEGIN 
                FOR tabname IN (SELECT table_name FROM information_schema.tables WHERE table_name LIKE 'DWD_IconD2_%') 
                LOOP 
                    EXECUTE 'DROP TABLE IF EXISTS ' || tabname; 
                END LOOP; 
            END $$;
        """)

        # Delete lat_lon_schema table
        cursor.execute("""
            DROP TABLE IF EXISTS lat_lon_schema;
        """)

        # Delete model table
        cursor.execute("""
            DROP TABLE IF EXISTS model;
        """)

        # Delete provider table
        cursor.execute("""
            DROP TABLE IF EXISTS provider;
        """)

        # Commit the changes
        conn.commit()

        # Close the cursor and the connection
        cursor.close()
        conn.close()

        print("Tables deleted successfully.")

    except Exception as e:
        print(f"Error: {e}")

if __name__ == "__main__":
    delete_tables()
```
