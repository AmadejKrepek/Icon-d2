```python
import csv

# Define the bounding box coordinates
lat_min = 45.1512
lat_max = 47.1512
lon_min = 12.1955
lon_max = 16.1955

# Read data from CSV file
input_file = "./output_data.csv"
filtered_lines = []

with open(input_file, "r") as file:
    csv_reader = csv.reader(file)
    header = next(csv_reader)
    filtered_lines.append(header)

    for row in csv_reader:
        lat, lon, temp = map(float, row)
        if lat_min <= lat <= lat_max and lon_min <= lon <= lon_max:
            filtered_lines.append(row)

# Write filtered data to a new CSV file
output_file = "filtered_data.csv"

with open(output_file, "w", newline="") as file:
    csv_writer = csv.writer(file)
    csv_writer.writerows(filtered_lines)

print("Filtered data has been saved to 'filtered_data.csv'")
```
