```python
import psycopg2
from dotenv import load_dotenv
import os
import csv

# Load environment variables from .env
load_dotenv()

def export_table_to_csv(table_name, output_directory):
    try:
        # Access the environment variables
        DB_USERNAME = os.getenv("DB_USERNAME")
        DB_PASSWORD = os.getenv("DB_PASSWORD")
        DB_HOST = os.getenv("DB_HOST")
        DB_PORT = os.getenv("DB_PORT")
        DB_NAME = os.getenv("DB_NAME")

        # Establish a connection to the PostgreSQL database
        conn = psycopg2.connect(
            host=DB_HOST,
            port=DB_PORT,
            database=DB_NAME,
            user=DB_USERNAME,
            password=DB_PASSWORD
        )

        # Create a cursor object
        cursor = conn.cursor()

        # Retrieve data from the specified table
        cursor.execute(f'SELECT * FROM "{table_name}"')
        rows = cursor.fetchall()

        # Define the output CSV file path
        csv_file_path = os.path.join(output_directory, f'{table_name}.csv')

        # Write data to the CSV file
        with open(csv_file_path, 'w', newline='') as csv_file:
            csv_writer = csv.writer(csv_file)
            # Write the header row based on the table columns
            column_names = [desc[0] for desc in cursor.description]
            csv_writer.writerow(column_names)
            # Write the data rows
            csv_writer.writerows(rows)

        print(f"Data from table '{table_name}' exported to CSV: {csv_file_path}")

        # Close the cursor and the connection
        cursor.close()
        conn.close()

    except Exception as e:
        print(f"Error: {e}")

def show_tables_and_data(limit_per_table=5):
    try:
        specific_table_name = 'total_precipitation'

        export_table_to_csv(specific_table_name, 'data')
    except Exception as e:
        print(f"Error: {e}")

if __name__ == "__main__":
    show_tables_and_data(limit_per_table=1)
```
