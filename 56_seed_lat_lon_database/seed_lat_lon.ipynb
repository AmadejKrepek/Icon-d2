```python
import psycopg2
import csv
from dotenv import load_dotenv
import os

# Load environment variables from .env
load_dotenv()

def seed_lat_lon_schema():
    try:
        # Access the environment variables
        DB_USERNAME = os.getenv("DB_USERNAME")
        DB_PASSWORD = os.getenv("DB_PASSWORD")
        DB_HOST = os.getenv("DB_HOST")
        DB_PORT = os.getenv("DB_PORT")
        DB_NAME = os.getenv("DB_NAME")

        # Database connection
        conn = psycopg2.connect(
            host=DB_HOST,
            port=DB_PORT,
            database=DB_NAME,
            user=DB_USERNAME,
            password=DB_PASSWORD
        )

        cursor = conn.cursor()

        # Path to CSV file containing latitude and longitude data for DWD
        csv_file_dwd = "../50_extract_lat_lon_from_csv/data/output_icon_d2_lat_lon.csv"
        
        # Define provider_id, model_id, and model_run for DWD data
        table_name = "lat_lon_schema"
        provider_id_dwd = "6be8cea2-f29b-4198-aa68-10c57845ad25"
        model_id_dwd = "581e4233-dc8c-44d3-b351-c115dc32fc53"
        
        # Create empty lists to collect latitude and longitude values
        dwd_latitudes = []
        dwd_longitudes = []
        
        # Insert data into lat_lon_schema for DWD
        with open(csv_file_dwd, "r") as file_dwd:
            csv_reader_dwd = csv.reader(file_dwd)
            next(csv_reader_dwd)  # Skip the header row if present
            for row in csv_reader_dwd:
                # Extract latitude and longitude values
                latitude = float(row[0])
                longitude = float(row[1])
                
                # Append latitude and longitude to the lists
                dwd_latitudes.append(latitude)
                dwd_longitudes.append(longitude)
        
        # Insert a single record for DWD with arrays of latitude and longitude
        cursor.execute("""
            INSERT INTO {} (provider_id, model_id, latitudes, longitudes)
            VALUES (%s, %s, %s, %s)
        """.format(table_name), (provider_id_dwd, model_id_dwd, dwd_latitudes, dwd_longitudes))
        
        # Commit the changes for DWD data
        conn.commit()
        
        # Clear the lists for ARSO data
        dwd_latitudes.clear()
        dwd_longitudes.clear()

        # Path to CSV file containing latitude and longitude data for ARSO
        csv_file_arso = "../50_extract_lat_lon_from_csv/data/output_aladin_lat_lon.csv"
        
        # Define provider_id, model_id, and model_run for ARSO data
        provider_id_arso = "66b97430-1782-4793-906d-f77b4303373b"
        model_id_arso = "62490a51-1f97-4858-874b-267ea06bdbcc"
        
        # Insert data into lat_lon_schema for ARSO
        with open(csv_file_arso, "r") as file_arso:
            csv_reader_arso = csv.reader(file_arso)
            next(csv_reader_arso)  # Skip the header row if present
            for row in csv_reader_arso:
                # Extract latitude and longitude values
                latitude = float(row[0])
                longitude = float(row[1])
                
                # Append latitude and longitude to the lists
                dwd_latitudes.append(latitude)
                dwd_longitudes.append(longitude)
        
        # Insert a single record for ARSO with arrays of latitude and longitude
        cursor.execute("""
            INSERT INTO {} (provider_id, model_id, latitudes, longitudes)
            VALUES (%s, %s, %s, %s)
        """.format(table_name), (provider_id_arso, model_id_arso, dwd_latitudes, dwd_longitudes))
        
        # Commit the changes for ARSO data
        conn.commit()

        # Close the cursor and the connection
        cursor.close()
        conn.close()

        print("Latitude and Longitude data seeded successfully.")

    except Exception as e:
        print(f"Error: {e}")

if __name__ == "__main__":
    seed_lat_lon_schema()
```
