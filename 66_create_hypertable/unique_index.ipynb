```python
import psycopg2
import os
from dotenv import load_dotenv

# Load environment variables from .env
load_dotenv()

# Database connection information
DB_USERNAME = os.getenv("DB_USERNAME")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")

def create_unique_index(table_name, timestamp_column):
    try:
        # Establish a connection to the PostgreSQL database
        conn = psycopg2.connect(
            host=DB_HOST,
            port=DB_PORT,
            database=DB_NAME,
            user=DB_USERNAME,
            password=DB_PASSWORD
        )

        # Create a cursor object
        cursor = conn.cursor()

        # Define the SQL statement to create a unique index on the timestamp_column
        create_index_sql = f"""
            CREATE UNIQUE INDEX IF NOT EXISTS {table_name}_{timestamp_column}_idx
            ON {table_name} ({timestamp_column});
        """

        # Execute the SQL statement to create the unique index
        cursor.execute(create_index_sql)

        conn.commit()

        print(f"Unique index creation for table '{table_name}' on column '{timestamp_column}' completed successfully.")

        # Close the cursor and the connection
        cursor.close()
        conn.close()

    except Exception as e:
        print(f"Error: {e}")

if __name__ == "__main__":
    # Specify the table name and timestamp column name to create the unique index
    table_name = "total_precipitation"  # Replace with your table name
    timestamp_column = "start_date"  # Replace with the timestamp column name

    create_unique_index(table_name, timestamp_column)
```
