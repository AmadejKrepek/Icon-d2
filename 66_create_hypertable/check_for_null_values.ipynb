```python
import psycopg2
import os
from dotenv import load_dotenv

# Load environment variables from .env
load_dotenv()

# Database connection information
DB_USERNAME = os.getenv("DB_USERNAME")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")

def create_hypertable(table_name, timestamp_column, t2):
    try:
        # Establish a connection to the PostgreSQL database
        conn = psycopg2.connect(
            host=DB_HOST,
            port=DB_PORT,
            database=DB_NAME,
            user=DB_USERNAME,
            password=DB_PASSWORD
        )

        # Create a cursor object
        cursor = conn.cursor()

        # Define the SQL statement to create a hypertable using TimescaleDB
        create_hypertable_sql = f"""
            CREATE EXTENSION IF NOT EXISTS timescaledb CASCADE;
            SELECT create_hypertable('{table_name}', '{timestamp_column}', chunk_time_interval => interval '1 day', migrate_data => true);
        """

        # Execute the SQL statement to create the hypertable
        cursor.execute(create_hypertable_sql)

        conn.commit()

        print(f"Hypertable creation for table '{table_name}' completed successfully.")

        # Close the cursor and the connection
        cursor.close()
        conn.close()

    except Exception as e:
        print(f"Error: {e}")

if __name__ == "__main__":
    # Specify the table name and timestamp column names to create the hypertable
    table_name = "total_precipitation"  # Replace with your table name
    timestamp_column = "end_date"
    t2 = "end_date"  # Replace with the timestamp column name used for partitioning

    create_hypertable(table_name, timestamp_column, t2)
```
