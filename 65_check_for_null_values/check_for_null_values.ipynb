```python
import psycopg2
import os
from dotenv import load_dotenv

# Load environment variables from .env
load_dotenv()

# Database connection information
DB_USERNAME = os.getenv("DB_USERNAME")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")

def check_null_values(table_name, column_name):
    try:
        # Establish a connection to the PostgreSQL database
        conn = psycopg2.connect(
            host=DB_HOST,
            port=DB_PORT,
            database=DB_NAME,
            user=DB_USERNAME,
            password=DB_PASSWORD
        )

        # Create a cursor object
        cursor = conn.cursor()

        # Execute a SQL query to count null values in the specified column
        cursor.execute(f"SELECT COUNT(*) FROM {table_name} WHERE {column_name} IS NULL")

        # Fetch the count
        count = cursor.fetchone()[0]

        print(f"Number of null values in column '{column_name}' of table '{table_name}': {count}")

        # Close the cursor and the connection
        cursor.close()
        conn.close()

    except Exception as e:
        print(f"Error: {e}")

if __name__ == "__main__":
    # Specify the table name and column name to check for null values
    table_name = "total_precipitation"  # Replace with your table name
    column_name = "data"  # Replace with the column name of interest

    check_null_values(table_name, column_name)
```
