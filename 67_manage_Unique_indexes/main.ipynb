```python
import psycopg2
import os
from dotenv import load_dotenv

# Load environment variables from .env
load_dotenv()

# Database connection information
DB_USERNAME = os.getenv("DB_USERNAME")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")

def list_indexes(table_name):
    try:
        # Establish a connection to the PostgreSQL database
        conn = psycopg2.connect(
            host=DB_HOST,
            port=DB_PORT,
            database=DB_NAME,
            user=DB_USERNAME,
            password=DB_PASSWORD
        )

        # Create a cursor object
        cursor = conn.cursor()

        # Define the SQL statement to retrieve indexes for a specific table
        list_indexes_sql = f"""
            SELECT indexname
            FROM pg_indexes
            WHERE tablename = '{table_name}';
        """

        # Execute the SQL statement to list indexes
        cursor.execute(list_indexes_sql)

        # Fetch all rows (indexes)
        indexes = cursor.fetchall()

        if indexes:
            print(f"Indexes for table '{table_name}':")
            for index in indexes:
                print(index[0])
        else:
            print(f"No indexes found for table '{table_name}'.")

        # Close the cursor and the connection
        cursor.close()
        conn.close()

    except Exception as e:
        print(f"Error: {e}")

if __name__ == "__main__":
    # Specify the table name for which you want to list indexes
    table_name = "total_precipitation"  # Replace with your table name
    list_indexes(table_name)
```
