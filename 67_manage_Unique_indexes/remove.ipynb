```python
import psycopg2
import os
from dotenv import load_dotenv

# Load environment variables from .env
load_dotenv()

# Database connection information
DB_USERNAME = os.getenv("DB_USERNAME")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")

def drop_index(table_name, index_name):
    try:
        # Establish a connection to the PostgreSQL database
        conn = psycopg2.connect(
            host=DB_HOST,
            port=DB_PORT,
            database=DB_NAME,
            user=DB_USERNAME,
            password=DB_PASSWORD
        )

        # Create a cursor object
        cursor = conn.cursor()

        # Define the SQL statement to drop the specified index
        drop_index_sql = f"""
            DROP INDEX IF EXISTS {index_name};
        """

        # Execute the SQL statement to drop the index
        cursor.execute(drop_index_sql)

        conn.commit()

        print(f"Index '{index_name}' dropped successfully from table '{table_name}'.")

        # Close the cursor and the connection
        cursor.close()
        conn.close()

    except Exception as e:
        print(f"Error: {e}")

if __name__ == "__main__":
    # Specify the table name and index name to be dropped
    table_name = "total_precipitation"  # Replace with your table name
    index_name = "total_precipitation_start_date_idx"  # Replace with the index name to be dropped
    drop_index(table_name, index_name)
```
