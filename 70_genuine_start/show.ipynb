```python
import psycopg2
import os
import csv
from dotenv import load_dotenv
import uuid

# Load environment variables from .env
load_dotenv()

# Database connection information
DB_USERNAME = os.getenv("DB_USERNAME")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")

def get_data_and_write_to_csv(table_name, output_file, provider_id, model_id):
    try:
        # Establish a connection to the PostgreSQL database
        conn = psycopg2.connect(
            host=DB_HOST,
            port=DB_PORT,
            database=DB_NAME,
            user=DB_USERNAME,
            password=DB_PASSWORD
        )

        # Create a cursor object
        cursor = conn.cursor()

        # Query the data from the specified table for the given provider_id and model_id
        cursor.execute(f"SELECT * FROM {table_name} WHERE provider_id = %s AND model_id = %s", (provider_id, model_id))

        # Fetch all rows
        rows = cursor.fetchall()

        if rows:
            # Get the column names
            column_names = [desc[0] for desc in cursor.description]

            # Write data to CSV file
            with open(output_file, "w", newline="") as csv_file:
                csv_writer = csv.writer(csv_file)
                csv_writer.writerow(column_names)  # CSV header
                csv_writer.writerows(rows)

            print(f"Data from table '{table_name}' for provider_id {provider_id} and model_id {model_id} written to '{output_file}' successfully.")
        else:
            print(f"No data found in table '{table_name}' for provider_id {provider_id} and model_id {model_id}.")

        # Close the cursor and the connection
        cursor.close()
        conn.close()

    except Exception as e:
        print(f"Error: {e}")

if __name__ == "__main__":
    # Specify the table name, output file name, provider_id (as a string), and model_id (as a string)
    table_name = "lat_lon_schema"  # Replace with your table name
    output_file = "./data/output_lat_lon.csv"  # Replace with your desired output file name
    provider_id = "6be8cea2-f29b-4198-aa68-10c57845ad25"  # Replace with the desired provider_id as a string
    model_id = "581e4233-dc8c-44d3-b351-c115dc32fc53"  # Replace with the desired model_id as a string

    get_data_and_write_to_csv(table_name, output_file, provider_id, model_id)
```
