```python
import csv

# Initialize a list to store coordinates until the first duplicate appears
unique_coordinates = []

icon_d2_path = './data/icon_d2_lat_lon.csv'
aladin_path = './data/aladin_lat_lon.csv'

icon_d2_output_path = './data/output_icon_d2_lat_lon.csv'
aladin_output_path = './data/output_aladin_lat_lon.csv'

# Read the CSV file
with open(icon_d2_path, 'r') as csv_file:
    csv_reader = csv.reader(csv_file)
    next(csv_reader)  # Skip the header row if present

    for row in csv_reader:
        # Extract latitude and longitude from the row
        latitude, longitude = float(row[0]), float(row[1])

        # Check if the current coordinate is the same as the first coordinate in the list
        if not unique_coordinates:
            # If the list is empty, add the current coordinate
            unique_coordinates.append([latitude, longitude])
        elif [latitude, longitude] != unique_coordinates[0]:
            # If it's not a duplicate, add it to the list
            unique_coordinates.append([latitude, longitude])
        else:
            # If a duplicate appears, break the loop
            break

# Write the unique coordinates to a new CSV file with a header
with open(icon_d2_output_path, 'w', newline='') as csv_file:
    csv_writer = csv.writer(csv_file)

    # Write the header row
    csv_writer.writerow(['Latitude', 'Longitude'])

    # Write each unique coordinate as a row in the output CSV file
    for coordinate in unique_coordinates:
        csv_writer.writerow(coordinate)
```
