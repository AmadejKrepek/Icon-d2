```python
import csv
from collections import defaultdict

# Input CSV file name
input_file = './data/aladin_lat_lon.csv'

# Output CSV file name
output_file = './data/aladin_lat_lon_grouped.csv'

# Set to store unique latitude and longitude combinations
unique_lat_lon = set()

try:
    # Read the input CSV file
    with open(input_file, 'r') as csv_file:
        reader = csv.DictReader(csv_file)

        # Assuming the columns are named 'Latitude' and 'Longitude'
        for row in reader:
            latitude = row['Latitude']
            longitude = row['Longitude']

            # Check if the latitude and longitude are not empty
            if latitude and longitude:
                # Create a unique key by concatenating latitude and longitude
                key = f"{latitude}_{longitude}"
                unique_lat_lon.add(key)

    # Create a new CSV file for unique latitude and longitude combinations
    with open(output_file, 'w', newline='') as csv_output_file:
        fieldnames = ['Latitude', 'Longitude']
        writer = csv.DictWriter(csv_output_file, fieldnames=fieldnames)

        # Write the header
        writer.writeheader()

        # Write unique latitude and longitude data
        for key in unique_lat_lon:
            latitude, longitude = key.split('_')
            writer.writerow({'Latitude': latitude, 'Longitude': longitude})

    print(f"Latitude and Longitude data grouped and written to {output_file}")
except FileNotFoundError:
    print(f"The file {input_file} does not exist.")
except Exception as e:
    print(f"An error occurred: {str(e)}")
```
