```python
import csv

# Input CSV file name
input_file = './data/icon_d2.csv'

# Output CSV file name
output_file = './data/icon_d2_lat_lon.csv'

# Lists to store latitude and longitude data
latitudes = []
longitudes = []

try:
    # Read the input CSV file
    with open(input_file, 'r') as csv_file:
        reader = csv.DictReader(csv_file)
        
        # Assuming the columns are named 'Latitude' and 'Longitude'
        for row in reader:
            latitudes.append(row['Latitude'])
            longitudes.append(row['Longitude'])

    # Create a new CSV file for latitude and longitude
    with open(output_file, 'w', newline='') as csv_output_file:
        fieldnames = ['Latitude', 'Longitude']
        writer = csv.DictWriter(csv_output_file, fieldnames=fieldnames)
        
        # Write the header
        writer.writeheader()
        
        # Write latitude and longitude data
        for lat, lon in zip(latitudes, longitudes):
            writer.writerow({'Latitude': lat, 'Longitude': lon})
    
    print(f"Latitude and Longitude data extracted and written to {output_file}")
except FileNotFoundError:
    print(f"The file {input_file} does not exist.")
except Exception as e:
    print(f"An error occurred: {str(e)}")
```
