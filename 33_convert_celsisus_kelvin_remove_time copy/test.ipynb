```python
import pandas as pd

# Read the CSV file into a DataFrame
csv_file = '../data/test/temperature/temperature_data_2km.csv'  # Replace with the actual file path
df = pd.read_csv(csv_file)

# Extract min and max latitude and longitude values
min_latitude = df['Latitude'].min()
max_latitude = df['Latitude'].max()
min_longitude = df['Longitude'].min()
max_longitude = df['Longitude'].max()

print(f"Minimum Latitude: {min_latitude}")
print(f"Maximum Latitude: {max_latitude}")
print(f"Minimum Longitude: {min_longitude}")
print(f"Maximum Longitude: {max_longitude}")
```
