```python
import matplotlib.font_manager

def find_similar_fonts(target_font_name):
    similar_fonts = []
    font_manager = matplotlib.font_manager.FontManager()
    
    for font in font_manager.ttflist:
        if target_font_name.lower() in font.name.lower():
            similar_fonts.append(font.name)
    
    return similar_fonts

target_font_name = "Neue Haas Grotesk Text Pro"
similar_fonts = find_similar_fonts(target_font_name)

if similar_fonts:
    print(f"Similar fonts to '{target_font_name}':")
    for font in similar_fonts:
        print(font)
else:
    print(f"No similar fonts found to '{target_font_name}'.")
```
