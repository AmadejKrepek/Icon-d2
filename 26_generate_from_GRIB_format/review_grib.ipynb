```python
import bz2
import pygrib
import tempfile
import os

# Specify the path to the compressed GRIB2 file
compressed_file_path = "./downloaded_grib_files/icon-d2_germany_icosahedral_single-level_2023081815_000_2d_t_2m.grib2.bz2"

# Open the compressed GRIB2 file using bz2
with bz2.open(compressed_file_path, "rb") as f:
    # Decompress the content and read it using pygrib
    decompressed_data = f.read()

    # Create a temporary file to store the decompressed data
    with tempfile.NamedTemporaryFile(delete=False) as temp_file:
        temp_file.write(decompressed_data)
        temp_file_path = temp_file.name

    # Open the decompressed data using pygrib
    grbs = pygrib.open(temp_file_path)

    # Loop through the messages in the GRIB file
    for grb in grbs:
        print("Parameter Name:", grb.parameterName)
        print("Level Type:", grb.levelType)
        print("Level:", grb.level)
        print("Valid Time:", grb.validityDate, grb.validityTime)
        print("Values Shape:", grb.values.shape)
        print("Values Min:", grb.values.min())
        print("Values Max:", grb.values.max())
        print("*******************************")

    # Close the GRIB file
    grbs.close()

# Clean up: Delete the temporary file
os.remove(temp_file_path)
```
