```python
import psycopg2
from dotenv import load_dotenv
import os

# Load environment variables from .env
load_dotenv()

def show_first_5_rows_of_table_and_save_to_file(table_name, output_file):
    try:
        # Access the environment variables
        DB_USERNAME = os.getenv("DB_USERNAME")
        DB_PASSWORD = os.getenv("DB_PASSWORD")
        DB_HOST = os.getenv("DB_HOST")
        DB_PORT = os.getenv("DB_PORT")
        DB_NAME = os.getenv("DB_NAME")

        # Establish a connection to the PostgreSQL database
        conn = psycopg2.connect(
            host=DB_HOST,
            port=DB_PORT,
            database=DB_NAME,
            user=DB_USERNAME,
            password=DB_PASSWORD
        )

        # Create a cursor object
        cursor = conn.cursor()

        # Retrieve the first 5 rows of the specified table
        cursor.execute(f'SELECT * FROM "{table_name}" LIMIT 2')
        rows = cursor.fetchall()

        # Display the first 5 rows of the table
        print(f"First 5 rows of Table: {table_name}")
        for row in rows:
            print(row)

        # Save the result to the specified output file
        with open(output_file, 'w') as file:
            for row in rows:
                file.write(','.join(map(str, row)) + '\n')

        # Close the cursor and the connection
        cursor.close()
        conn.close()

        print(f"Result saved to {output_file}")

    except Exception as e:
        print(f"Error: {e}")

if __name__ == "__main__":
    # Specify the table name you want to show the first 5 rows of
    table_name_to_show = "model"
    output_file_name = "output.txt"
    show_first_5_rows_of_table_and_save_to_file(table_name_to_show, output_file_name)
```
